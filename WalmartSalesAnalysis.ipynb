```python
df['date'] = pd.to_datetime(df['date'], format='%d-%m-%Y')

df['day_name'] = df['date'].dt.day_name()
df['month_name'] = df['date'].dt.month_name()
df['year'] = df['date'].dt.year
```

```python
df[['date','day_name','month_name','year']].head()
```

```python
def get_shift(time):
    hour = pd.to_datetime(time).hour

    if hour < 12:
        return 'Morning'
    elif hour < 17:
        return 'Afternoon'
    else:
        return 'Evening'

df['shift'] = df['time'].apply(get_shift)
```

```python
df[['time','shift']].head()
```

```python
df.to_csv('walmart_feature_engineered.csv', index=False)
```
